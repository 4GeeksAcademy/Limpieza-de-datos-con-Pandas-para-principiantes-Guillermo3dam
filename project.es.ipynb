#### Ejercicio 01. ¿Cuál es la casa más cara en todo el dataset? (★☆☆)

Imprime la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle del Prado, Nº20 es la más cara y su precio es de 5000000 USD`

```python

house3= ds.iloc[2]
# print(house3)  # imprimo una fila al azar para saber los nombres de las columnas

most_expensive = ds.loc[ds["price"].idxmax()]
print(f"La casa con direccion {most_expensive['address']} es la más cara y su precio es de {most_expensive['price']} USD") # dentro de un f string no podemos usar mas veces las dobles comillas
```

#### Ejercicio 02. ¿Cuál es la casa más barata del dataset? (★☆☆)

Este ejercicio es similar al ejercicio anterior, solo que ahora buscamos la casa con el precio más bajo. Recuerda imprimir la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle Alcalá, Nº58 es la más barata y su precio es de 12000 USD`

```python

ds = ds[ds["price"] > 0] # ya que muestra casas que valen 0 actualizamos por si hay fallos
lowest_expensive = ds.loc[ds["price"].idxmin()]
print(f"La casa con direccion {lowest_expensive['address']} es la más barata y su precio es de {lowest_expensive['price']} USD") # dentro de un f string no podemos usar mas veces las dobles comillas
```

#### Ejercicio 03. ¿Cuál es la casa más grande y la más pequeña del dataset? (★☆☆)

Imprime la dirección y el área de las casas seleccionadas. Por ejemplo:

`La casa más grande está ubicada en Calle Gran Vía, Nº38 y su superficie es de 5000 metros`

`La casa más pequeña está ubicada en Calle Mayor, Nº12 y su superficie es de 200 metros`

Este ejercicio es similar al anterior, pero buscamos las casas más grandes y pequeñas en base a su superficie.

```python

biggest_house = ds.loc[ds["surface"].idxmax()]
print(f"La casa más grande está ubicada en {biggest_house['address']} y su superficie es de {biggest_house['surface']} metros")

smallest_house = ds.loc[ds["surface"].idxmin()]
print(f"La casa más pequeña está ubicada en {smallest_house['address']} y su superficie es de {smallest_house['surface']} metros")
```

#### Ejercicio 07. ¿Cuál es la media de precios en la población de "Arroyomolinos (Madrid)"? (★★☆)

Imprima el valor obtenido de la columna level5.

```python
subset = ds[ds["level5"] == "Arroyomolinos (Madrid)"] # guardo un nuevo ds con las ubicaciones de arroyomolinos

mean_price = subset["price"].mean() # guarda la media de la columna precios
print(mean_price)
```

#### Ejercicio 08. Trazar el histograma de los precios para la población de "Arroyomolinos (Madrid)" y explica qué observas (★★☆)

Imprime el histograma de los precios y escribe en la celda del Markdown un breve análisis del trazado.


```python
import matplotlib.pyplot as plt

# Trazar el histograma de precios
prices = subset["price"]
plt.hist(prices, bins=15)
plt.title('Precio historico Arroyomolinos (Madrid)')
plt.xlabel('Precio')
plt.ylabel('Numero de propietarios')
plt.grid(axis='y')

plt.show
```

La mayoria de precios estan como es de esperar en el centro, algunos mas lujosos se disparan pero no es lo normal, y muestra uno con muy poco precio que quiza sea un error
